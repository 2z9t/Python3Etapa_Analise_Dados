```python
!pip install pandas

import pandas as pd

df = pd.read_excel("CD2022_Populacao_2010_Compatibilizada_20231222.xlsx",
                   sheet_name="Municípios", header=4)

df = df.dropna(subset=["UF", "NOME DO MUNICÍPIO"])
df.head()
```

```python
estado = df.groupby("UF")[["População 2010 (Alterações de Limites até 2022)1",
                            "População Censo 2022"]].sum().reset_index()

estado["DIFERENCA"] = estado["População Censo 2022"] - estado["População 2010 (Alterações de Limites até 2022)1"]

estado = estado.sort_values("DIFERENCA", ascending=False)

estado
```

```python
estado.to_csv("estado.csv", sep=";", index=False)
```

```python
municipio = df[["UF", "NOME DO MUNICÍPIO",
                "População 2010 (Alterações de Limites até 2022)1",
                "População Censo 2022"]].copy()

municipio["DIFERENCA"] = municipio["População Censo 2022"] - municipio["População 2010 (Alterações de Limites até 2022)1"]

municipio = municipio.sort_values("DIFERENCA", ascending=False)

municipio
```

```python
municipio.to_csv("municipio.csv", sep=";", index=False)
```
